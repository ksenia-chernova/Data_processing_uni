## Задание 1

Данные - текстовые отзывы на фильмы, разделенные на негативные и позитивные. Что мы собираемся с ними сделать:

Выделить отдельные слова, привести из к единому виду (убрав словоформы);
Построить вектор из всех слов, а потом для каждого сообщения заполнить вектор такой же длины нулями или единицами в зависимости от того, было ли слово в тексте;
Объединить все, разделить данные на обучающую и тестовую выборки;
Создать классификатор и обучить его.

В папке с лабораторной скачайте файл data.zip.

В цикле считайте из папки neg txt’шники в list строк. В тот же list добавьте строки из папки pos. Параллельно создайте list со значениями 0 для негативных отзывов и 1 – для положительных.

```python
import os

file_path_pos = "/content/drive/MyDrive/01_lab_data/data/movie/pos"
file_path_neg = "/content/drive/MyDrive/01_lab_data/data/movie/neg"

lines_from_txt = []
labels = []

def read_txt_func(path, label):
    for filename in os.listdir(path):
        file_path = os.path.join(path, filename)
        with open (file_path, 'r', encoding='utf-8') as file:
            lines = file.readlines()
            text = ''.join(lines)
            lines_from_txt.append(text)
            labels.append(label)

read_txt_func(file_path_pos, 1)
read_txt_func(file_path_neg, 0)

print(f"Всего отзывов: {len(lines_from_txt)}")
print(f"Негативных (0): {labels.count(0)}")
print(f"Позитивных (1): {labels.count(1)}")
```

```python
all_texts = ' '.join(lines_from_txt)
unique_words = list(set(all_texts.split()))

print(f"количество уникальных слов: {len(unique_words)}")
```

Выполнить прямое кодирование списков слов в векторы нулей и единиц (в numpy массив). Это может означать, например, преобразование последовательности [good, day] в 10 000-мерный вектор, все элементы которого содержат нули, кроме элементов с индексами 3 и 5 (номеров этих слов в словаре), которые содержат единицы. Затем их можно передать на вход модели машинного обучения.

```python
results = np.zeros((len(lines_from_txt), len(unique_words)))

for i, text in enumerate(lines_from_txt):
    for word in text.split():
        if word in unique_words:
            results[i, unique_words.index(word)] = 1.

print(f"\nРазмер матрицы: {results.shape}")
print("Первые 20 строк матрицы (первые 10 признаков):")
print(results[:20, :10])
```

После векторизации нужно разбить выборку на обучающую и тестовую.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(results,
                                                    labels,
                                                    test_size=0.2,
                                                    random_state=0)
```

Создать и обучить классификатор

```python
from sklearn.naive_bayes import BernoulliNB

clf = BernoulliNB()
clf.fit(X_train, Y_train)
```

Посмотреть, что выдает классификатор на тестовой выборке и сравнить с правильными значениями.

```python
from sklearn import metrics
from sklearn.metrics import classification_report, confusion_matrix

print(f"Точность на тестовой выборке: {metrics.accuracy_score(clf.predict(X_test), Y_test):.4f}")

print("\nМатрица ошибок:")
print(confusion_matrix(Y_test, clf.predict(X_test)))

print("\nОтчет по классификации:")
print(classification_report(Y_test, clf.predict(X_test)))
```

```python
import numpy as np
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

X_train_norm = X_train.copy()
X_test_norm = X_test.copy()

mean = X_train_norm.mean(axis=0)
std = X_train_norm.std(axis=0)

std[std == 0] = 1

X_train_norm -= mean
X_train_norm /= std

X_test_norm -= mean
X_test_norm /= std

ridge_model = Ridge(alpha=1.0)
ridge_model.fit(X_train_norm, Y_train)

y_pred = ridge_model.predict(X_test_norm)

mse = mean_squared_error(Y_test, y_pred)
print(f"\nСреднеквадратичная ошибка (MSE) после нормализации: {mse:.4f}")

ridge_no_norm = Ridge(alpha=1.0)
ridge_no_norm.fit(X_train, Y_train)
mse_no_norm = mean_squared_error(Y_test, ridge_no_norm.predict(X_test))
print(f"\nMSE без нормализации: {mse_no_norm:.4f}")
print(f"Улучшение после нормализации: {mse_no_norm - mse:.4f}")
```
